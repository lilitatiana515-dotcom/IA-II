El despertar de la memoria
El tema principal es el uso de redes neuronales recurrentes (RNN) para trabajar con datos secuenciales, como series de tiempo, y las técnicas desarrolladas para solucionar sus problemas de memoria.

1. Redes neuronales recurrentes
Una red neuronal recurrente (RNN) es una red que tiene conexiones hacia atrás, lo que le permite conservar información de estados anteriores y utilizarla para procesar secuencias de diferentes longitudes.

La red mantiene un estado oculto, que se actualiza a medida que recibe nuevos datos. Al desplegar una neurona recurrente a través del tiempo, se puede observar cómo la información pasa de un momento al siguiente.

Para entrenar estas redes se utiliza la retropropagación a través del tiempo (BPTT). Su objetivo es ajustar los parámetros de la red para reducir la función de pérdida, que indica qué tan diferente fue la predicción de la realidad.

Pasos de BPTT:
Pase hacia adelante: la red procesa la secuencia y genera una salida.

Evaluación de la salida: se calcula la función de pérdida.

Propagación hacia atrás: el error se transmite hacia los pasos anteriores.

Descenso del gradiente: se actualizan los parámetros para mejorar las predicciones.

2. Pronóstico de series de tiempo
Las redes recurrentes pueden utilizarse para realizar pronósticos de series temporales, es decir, predecir valores futuros a partir de información pasada.

Los modelos pueden ser:

Univariantes: utilizan una sola variable, por ejemplo, la cantidad de pasajeros de autobús.

Multivariantes: utilizan varias variables, como pasajeros de autobús, pasajeros de tren y el día.

También es posible realizar predicciones de varios pasos hacia adelante, por ejemplo, pronosticar 14 pasos futuros de manera progresiva.

3. El problema de la memoria corta
Las redes recurrentes simples tienen dificultades para recordar información que ocurrió muchos pasos atrás. Cuando una secuencia supera aproximadamente los 10 pasos, la red puede comenzar a perder información relevante.

Para solucionar este problema surgieron estructuras como:

LSTM (Long Short-Term Memory): utiliza una celda de memoria que permite conservar o eliminar información de forma controlada.

GRU (Gated Recurrent Unit): tiene una estructura más sencilla y utiliza compuertas para controlar la información que se conserva.

LSTM
Una celda LSTM cuenta con diferentes compuertas matemáticas que controlan el flujo de información. Entre ellas se encuentran:

Puerta de entrada: controla qué información nueva se incorpora.

Puerta de salida: controla qué información se utiliza como salida.

También existe un mecanismo para controlar qué información debe conservarse o descartarse.

GRU
La GRU simplifica la arquitectura de la LSTM y utiliza menos mecanismos internos. Su objetivo también es controlar qué información conservar y cuál actualizar, permitiendo manejar dependencias más largas.

4. WaveNet y CNN
Cuando las secuencias tienen miles de pasos, las redes recurrentes pueden resultar difíciles de entrenar o procesar. Para estos casos también pueden utilizarse redes convolucionales (CNN) y arquitecturas como WaveNet.

WaveNet utiliza capas convolucionales organizadas de forma apilada para procesar secuencias largas.

Un concepto importante es la tasa de dilatación, que permite que las capas observen información cada vez más distante sin tener que utilizar una cantidad excesiva de capas.

También utiliza relleno causal, lo que garantiza que al realizar una predicción solo se utilice información disponible anteriormente y no información futura.

Idea principal
Las RNN permiten trabajar con secuencias porque mantienen información de estados anteriores. Sin embargo, tienen problemas para recordar información lejana. Las LSTM y GRU fueron diseñadas para mejorar esta memoria, mientras que arquitecturas como WaveNet y las CNN permiten procesar secuencias muy largas utilizando convoluciones y dilataciones.
