Aprendizaje por Conjuntos (Ensemble Learning)


1. Introducción a los Conjuntos
Origen: Se fundamenta en la idea humana de que la combinación de múltiples opiniones o modelos produce mejores decisiones.

Base teórica: Apoyado en la Ley de los Grandes Números, donde la agregación de múltiples predictores independientes tiende a reducir el error general.

2. Clasificadores por Votación
Mecanismo para combinar las predicciones de distintos modelos base:

Votación dura (Hard Voting): El resultado final es la clase que obtiene la mayoría matemática de votos de los clasificadores.

Votación blanda (Soft Voting): Se promedian las probabilidades estimadas por cada modelo y se selecciona la clase con la probabilidad promedio más alta.

3. Bagging, Pasting y OOB
Estrategias de muestreo para entrenar modelos en paralelo sobre distintos subconjuntos de datos:

Bagging: Muestreo aleatorio con reemplazo de las instancias de entrenamiento.

Pasting: Muestreo aleatorio sin reemplazo.

Entrenamiento paralelo: Todos los modelos del conjunto se entrenan simultáneamente en sus respectivos subconjuntos.

Instancias Fuera de la Bolsa (Out-Of-Bag / OOB): Datos que no fueron seleccionados en el muestreo de Bagging, utilizados directamente para la evaluación y validación del conjunto sin necesidad de un conjunto de prueba separado.

4. Bosques Aleatorios y Variantes
Estrategias de muestreo:

Parches aleatorios: Muestrea tanto instancias (filas) como características (columnas).

Subespacios aleatorios: Muestrea únicamente las características (columnas), manteniendo todas las instancias.

Árboles Extra (Extra-Trees): Variante extremadamente aleatorizada de los bosques aleatorios que introduce aleatoriedad adicional en los umbrales de división de los nodos para reducir la varianza.

5. Algoritmos de Boosting
Estrategia secuencial donde cada nuevo modelo intenta corregir los errores cometidos por los clasificadores anteriores:

AdaBoost (Adaptive Boosting)
Paso 1: Entrenar un modelo base inicial.

Paso 2: Aumentar el peso de las instancias mal clasificadas.

Paso 3: Entrenar un nuevo clasificador enfocado en los datos con mayor peso.

Paso 4: Repetir el ciclo secuencialmente hasta completar los clasificadores definidos.

Otras variantes de Boosting:
Gradient Boosting: Entrena nuevos modelos ajustándose a los residuos (errores) cometidos por los modelos anteriores mediante el descenso de gradiente.

Hist-Based Gradient Boosting: Optimización de Gradient Boosting que agrupa las características continuas en bins (contenedores discretos), acelerando drásticamente el entrenamiento en grandes volúmenes de datos.

6. Stacking (Stacked Generalization)
Concepto: En lugar de combinar predicciones mediante reglas fijas (como la votación), se entrena un meta-modelo denominado mezclador (blender).

Funcionamiento: El blender toma como entradas las predicciones realizadas por los modelos base y aprende a combinarlas para generar la predicción final óptima.
