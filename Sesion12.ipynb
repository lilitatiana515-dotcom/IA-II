1. ¿Qué son?
Los árboles de decisión son modelos de caja blanca, porque sus decisiones son transparentes y fáciles de interpretar. El modelo toma decisiones paso a paso mediante reglas del tipo:

¿La característica es menor o mayor que determinado valor?

La estructura comienza en un nodo raíz, continúa por nodos de ramificación y termina en nodos hoja, donde se obtiene la predicción.


2. Estructura del árbol
Un árbol está formado principalmente por:

Nodo raíz: punto donde comienza la decisión.

Nodos de ramificación: realizan preguntas o condiciones sobre los datos.

Ramas: representan las diferentes respuestas a esas condiciones.

Nodos hoja: contienen el resultado o predicción final.

Por ejemplo, utilizando el conjunto de datos Iris, el árbol puede dividir los datos según características de las flores hasta determinar a qué clase pertenece una observación.

3. ¿Cómo realiza una predicción?
El proceso puede resumirse en cuatro pasos:

Inicio: el dato entra por el nodo raíz.

Evaluación: se evalúa una condición.

Descenso: dependiendo del resultado, se sigue una rama.

Resultado: se llega a un nodo hoja y se obtiene la predicción.

Las divisiones generan fronteras de decisión ortogonales, es decir, separaciones que normalmente son perpendiculares a los ejes de las características.

4. CART e impureza
CART significa Classification and Regression Trees (árboles de clasificación y regresión).

Para decidir cómo dividir los datos, el árbol busca reducir la impureza, que representa qué tan mezcladas o desordenadas están las clases dentro de un grupo.

Dos medidas habituales son:

Gini: mide la probabilidad de clasificar incorrectamente un elemento si se asigna una clase según la distribución del nodo.

Entropía: mide el nivel de desorden o incertidumbre de los datos.

El árbol utiliza un proceso codicioso (greedy): en cada paso busca la división que parece más conveniente en ese momento.

5. Regularización y sobreajuste
Un árbol tiene mucha libertad para crear divisiones. Si no ponemos límites, puede llegar a memorizar los datos de entrenamiento, produciendo un fenómeno llamado sobreajuste (overfitting).

Para evitarlo se utiliza la regularización, por ejemplo:

Limitar la profundidad máxima del árbol.

Exigir un número mínimo de muestras para dividir un nodo.

Exigir un número mínimo de muestras en las hojas.

Podar el árbol.

Por eso podemos diferenciar entre:

Árbol sin restricciones: puede crecer demasiado y sobreajustarse.

Árbol regularizado: tiene límites para controlar su complejidad.

6. Árboles de regresión
Los árboles no solamente sirven para clasificación. También pueden utilizarse para regresión, cuando queremos predecir un valor continuo.

Por ejemplo, podrían utilizarse para predecir un precio, una temperatura o una cantidad.

En un árbol de regresión, una hoja normalmente predice un valor relacionado con las observaciones que quedaron dentro de ella, habitualmente el promedio de los datos del grupo.

En CART, las divisiones se buscan tratando de minimizar el error cuadrático.

7. Limitaciones de los árboles
Los árboles de decisión tienen algunas limitaciones importantes:

Son sensibles a la rotación y orientación de los datos. Una transformación de las características puede producir un árbol bastante diferente.

Tienen alta varianza. Pequeños cambios en los datos de entrenamiento pueden producir un árbol diferente.

Pueden sobreajustarse si no se controla su complejidad.

8. ¿Cómo superar la alta varianza?
La principal forma de reducir la alta varianza de los árboles es utilizar métodos de ensamble (ensemble methods).

En lugar de depender de un único árbol, se combinan muchos árboles para obtener una predicción más estable.

Un ejemplo muy importante es:

**Random Forest**
Un Random Forest construye muchos árboles de decisión y combina sus resultados.

La idea fundamental es:

Un árbol puede ser muy variable, pero combinar muchos árboles puede producir un modelo mucho más estable.

Otros métodos de ensamble también pueden utilizar árboles, como bagging y métodos de boosting.
