 Perceptrón y Redes Neuronales con Keras
1. De la biología a las redes neuronales
Las redes neuronales artificiales están inspiradas en el funcionamiento de las neuronas biológicas.

Una neurona biológica tiene principalmente:

Dendritas: reciben información.

Cuerpo celular: procesa la información.

Axón: transmite la señal a otras neuronas.

Sinapsis: conexión entre neuronas.

En 1943, se realizó una traducción matemática de este funcionamiento. En las redes neuronales artificiales:

Dendritas → entradas (inputs)
Sinapsis → pesos matemáticos (weights)
Neurona → operación matemática

2. El perceptrón
El perceptrón es uno de los modelos más básicos de una neurona artificial.

Recibe varias entradas, las multiplica por sus pesos y calcula una suma. Después aplica una función de activación.

De forma simplificada:

Entradas → pesos → suma → función de activación → salida

El perceptrón puede utilizar una unidad lógica de umbral (TLU), que utiliza una función escalón:

Si el resultado supera cierto umbral → 1

Si no lo supera → 0

En 1969, se estudiaron las limitaciones del perceptrón simple, especialmente para problemas como XOR. Un solo perceptrón no puede resolver determinados problemas que no son linealmente separables.

3. Perceptrón multicapa (MLP)
Para resolver problemas más complejos se utilizan los perceptrones multicapa (MLP).

Un MLP tiene:

Capa de entrada

Una o más capas ocultas

Capa de salida

Las capas ocultas permiten que la red aprenda relaciones más complejas.

Una representación sencilla sería:

Entrada → Capa oculta → Capa oculta → Salida

4. Retropropagación (Backpropagation)
La retropropagación es un método utilizado para entrenar redes neuronales.

El proceso puede resumirse en cuatro pasos:

Propagación hacia adelante: los datos pasan por la red y se obtiene una predicción.

Medir el error: se compara la predicción con el resultado correcto.

Propagación hacia atrás: el error se transmite hacia las capas anteriores.

Ajustar los pesos: se modifican los pesos para reducir el error.

Idea clave para memorizar:

Adelante → calcular error → atrás → ajustar pesos

5. Redes neuronales con Keras
Keras es una API de alto nivel utilizada para construir y entrenar redes neuronales.

Permite crear modelos de diferentes maneras.

API secuencial
La API Sequential sirve para construir redes donde las capas están conectadas de forma lineal.

Ejemplo conceptual:

Entrada → Capa → Capa oculta → Salida

En el ejemplo de Fashion-MNIST, se utilizan imágenes de prendas de vestir para entrenar una red neuronal.

Al construir el modelo pueden aparecer:

Capa de entrada/Flatten: transforma los datos de la imagen en un vector.

Capas ocultas Dense: contienen neuronas completamente conectadas.

ReLU: función de activación utilizada frecuentemente en las capas ocultas.

Softmax: función utilizada normalmente en la capa de salida para clasificación multiclase.

6. API Funcional de Keras
La API funcional permite crear arquitecturas más complejas que la API secuencial.

A diferencia de una estructura estrictamente lineal, permite tener:

Múltiples entradas.

Múltiples salidas.

Rutas que se dividen.

Rutas que se vuelven a conectar.

Un ejemplo es una red Wide & Deep, que combina diferentes rutas de procesamiento.

Idea clave:

Sequential = estructura lineal
Functional = estructuras complejas

7. API de subclases
La API de subclases permite crear modelos personalizados mediante código.

Se puede:

Heredar de la clase Model.

Definir las capas manualmente.

Crear un comportamiento dinámico.

Controlar con mayor libertad cómo funciona la red.

Es la opción más flexible de las tres.

8. Comparación de las tres API
API	Característica principal	Complejidad
Sequential	Capas conectadas linealmente	Baja
Functional	Arquitecturas con múltiples rutas	Alta
Subclassing	Modelo personalizado y dinámico	Muy alta
