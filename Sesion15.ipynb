**RESUMEN : Orden del Caos**

El aprendizaje no supervisado busca encontrar patrones y estructuras dentro de datos que no están previamente etiquetados. Una forma sencilla de entenderlo es pensar que el aprendizaje no supervisado es el pastel y el supervisado es el glaseado: primero se descubren las estructuras ocultas en los datos y luego se pueden aplicar técnicas supervisadas para aprovecharlas.

1. Agrupamiento o clustering
El agrupamiento consiste en identificar datos similares y asignarlos a diferentes grupos según sus características. Es una herramienta importante para la toma de decisiones, porque permite descubrir patrones que no eran evidentes inicialmente.

Uno de los algoritmos más conocidos es K-Means, que divide los datos en un número determinado de grupos o clusters. Para evaluar qué tan bien están separados los grupos puede utilizarse la puntuación de silueta.

2. Segmentación y aprendizaje semi-supervisado
La segmentación permite dividir los datos en grupos con características similares. En este contexto aparecen los datos semi-supervisados, donde se combina información etiquetada con información que no tiene etiquetas.

El aprendizaje activo sigue un proceso iterativo:

Entrenar → etiquetar → iterar → volver a entrenar.

Esto permite mejorar progresivamente el modelo utilizando las etiquetas más relevantes.

3. DBSCAN
DBSCAN es un algoritmo de agrupamiento basado en la densidad de los datos. A diferencia de K-Means, puede identificar grupos de formas complejas o no continuas y también puede ayudar a detectar datos que no pertenecen claramente a ningún grupo.

4. Anomalías y novedades
El análisis de anomalías busca identificar comportamientos o datos que se alejan de lo normal.

Anomalía: evento inusual que puede ser destructivo, problemático o indicar un riesgo.

Novedad: instancia poco frecuente que, aunque sea rara, puede resultar útil o representar una oportunidad.

Esto permite diferenciar entre datos que deben ser investigados por representar un problema y datos poco comunes que pueden aportar valor.

5. Mezclas gaussianas
Las mezclas gaussianas (Gaussian Mixture Models) permiten modelar los datos como una combinación de diferentes distribuciones. Pueden utilizarse para identificar grupos y aislar anomalías observando los contornos y la distribución de los datos.

Para determinar el número adecuado de grupos en estos modelos probabilísticos pueden utilizarse criterios como:

BIC (Bayesian Information Criterion)

AIC (Akaike Information Criterion)

Estas métricas ayudan a seleccionar una cantidad de grupos que represente adecuadamente los datos sin crear un modelo innecesariamente complejo.

Idea central
En conjunto, estas técnicas buscan transformar datos aparentemente desordenados en información útil para tomar decisiones. El proceso puede resumirse así:

Datos sin estructura → descubrir patrones → agrupar/segmentar → detectar anomalías → interpretar los resultados → generar valor.

La pregunta fundamental que queda al final es:

¿Cómo podemos desbloquear valor a partir de estos datos?

La respuesta está en utilizar el aprendizaje no supervisado para descubrir estructuras, comportamientos y oportunidades ocultas que posteriormente puedan convertirse en decisiones o acciones de negocio.
