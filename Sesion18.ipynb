*TensorFlow a medida*
1. Bloques fundamentales

TensorFlow trabaja principalmente con tensores, que son estructuras similares a los arreglos de NumPy y permiten almacenar y procesar datos numéricos.

Los arreglos de NumPy pueden trabajar con diferentes precisiones, como 64 bits.
TensorFlow suele utilizar 32 bits de precisión para trabajar de manera eficiente con redes neuronales.
Los tensores estándar son inmutables, es decir, una vez creados no se modifican directamente.
Para almacenar valores que necesitan cambiar durante el entrenamiento se utilizan las tf.Variable, que permiten mantener y actualizar valores en memoria.
2. Personalización avanzada

TensorFlow permite personalizar componentes importantes del entrenamiento de una red neuronal:

Función de pérdida: mide qué tan diferente es la predicción de la red respecto al resultado esperado.
Se puede crear una función de pérdida personalizada utilizando la clase tf.keras.losses.Loss y su método call().
Métricas: permiten evaluar el rendimiento del modelo durante el entrenamiento.
Las métricas personalizadas pueden crearse mediante tf.keras.metrics.Metric y actualizarse mediante update_state().
3. Cálculo de gradientes

TensorFlow utiliza la autodiferenciación, especialmente mediante el método de modo inverso, para calcular los gradientes de una red neuronal.

Es muy eficiente cuando existen muchos parámetros.
Utiliza la regla de la cadena para calcular las derivadas desde el resultado final hacia las variables que participaron en el cálculo.
tf.GradientTape registra las operaciones matemáticas realizadas y permite calcular posteriormente los gradientes.
Estos gradientes son utilizados para actualizar los parámetros de la red durante el entrenamiento.
4. Bucles de entrenamiento

Un bucle de entrenamiento personalizado permite controlar manualmente cada etapa del entrenamiento.

El proceso básico es:

Obtener un lote de datos → paso hacia adelante → calcular la pérdida → calcular gradientes → optimizar/actualizar parámetros.

La principal ventaja es la flexibilidad, ya que permite controlar con mayor precisión cómo se entrena el modelo e incluso asignar determinadas operaciones a diferentes procesadores o dispositivos.

5. Funciones y grafos

Un grafo de computación representa las operaciones matemáticas de un programa como una red de nodos y conexiones.

Sus ventajas incluyen:

Organizar las operaciones matemáticas.
Optimizar la ejecución.
Evitar operaciones o nodos que no sean necesarios.
Facilitar una ejecución más eficiente.

TensorFlow puede generar grafos automáticamente mediante AutoGraph y el trazado (tracing).

La función tf.function permite convertir funciones de Python en funciones que TensorFlow puede representar y ejecutar mediante grafos.

Al trabajar con tf.function es importante tener en cuenta que:

Se deben evitar dependencias innecesarias de librerías externas.
Las variables generalmente deben crearse en la primera ejecución o antes de utilizar la función.
Es recomendable trabajar con tensores.
Proporciona una forma de trabajar con una API de bajo nivel, dando mayor control sobre el funcionamiento del modelo.

**Resumen: Visión por computadora**

La visión por computadora busca que los computadores puedan analizar, reconocer e interpretar imágenes y videos. Las redes neuronales convolucionales (CNN) son una de las principales herramientas utilizadas para esta tarea.

1. Inspiración biológica

Las redes convolucionales están inspiradas en el funcionamiento del sistema visual biológico.

Los estudios realizados sobre las cortezas visuales de los gatos mostraron que algunas neuronas responden a características específicas de una imagen.
Por ejemplo, determinadas neuronas pueden reaccionar ante líneas, bordes u orientaciones específicas.
De aquí surge el concepto de campo receptivo local: es la región del campo visual en la que una neurona responde ante determinados estímulos.

En una CNN, esto se relaciona con que cada filtro analiza pequeñas regiones de la imagen en lugar de procesarla toda de una vez.

2. Capas y filtros

Las CNN utilizan filtros o kernels para detectar características dentro de una imagen.

El proceso básico es:

Definir filtro → aplicar zancada (stride) → agregar relleno (padding) → obtener mapa de características.

Filtro/Kernels: pequeñas matrices que recorren la imagen buscando características como bordes, líneas o texturas.
Stride (zancada): indica cuántos píxeles avanza el filtro cada vez que se mueve.
Padding (relleno): agrega valores, normalmente ceros, alrededor de la imagen para controlar el tamaño de la salida y conservar información de los bordes.
Mapas de características: cada filtro produce un mapa que indica dónde se encuentran las características que ese filtro detecta.

Cuando se utilizan varios filtros, se generan múltiples mapas de características que pueden representarse como una estructura 3D.

 Importante: el stride grande puede reducir la dimensión espacial. El padding no siempre reduce la dimensión; de hecho, puede ayudar a conservarla.

3. Agrupación o Pooling

A medida que una CNN tiene más capas, la cantidad de información puede aumentar considerablemente. El pooling permite reducir el tamaño de los datos.

Sus principales objetivos son:

Reducir las dimensiones de los datos.
Disminuir el uso de memoria y el costo computacional.
Conservar las características más importantes.
Hacer que la red sea menos sensible a pequeños cambios en la posición de una característica.

El ejemplo más conocido es el Max Pooling:

Divide la imagen o mapa de características en pequeñas regiones.
Selecciona únicamente el valor máximo de cada región.
De esta manera, conserva las características más destacadas mientras reduce el tamaño de los datos.

4. Arquitecturas famosas de CNN

Las redes convolucionales han evolucionado con el tiempo:

Año	Arquitectura	Característica
1998	LeNet-5	Una de las primeras CNN exitosas
2012	AlexNet	Impulsó el uso de CNN profundas para reconocimiento de imágenes
2014	GoogLeNet	Introdujo una arquitectura más eficiente mediante módulos Inception
2015	ResNet	Utilizó conexiones residuales para facilitar redes muy profundas
—	EfficientNet	Busca una buena relación entre precisión y eficiencia
—	MobileNet	Diseñada para dispositivos con recursos limitados

EfficientNet y MobileNet buscan obtener buenos resultados utilizando recursos de manera eficiente, por lo que son útiles cuando existen limitaciones de memoria o capacidad de procesamiento.

5. Modelos preentrenados y aprendizaje por transferencia

El aprendizaje por transferencia (Transfer Learning) consiste en utilizar un modelo que ya fue entrenado con una gran cantidad de imágenes y adaptarlo a una nueva tarea.

En lugar de entrenar una red completamente desde cero:

Se utiliza un modelo preentrenado.
Se mantienen muchas de las capas que ya aprendieron características generales.
Se modifican o ajustan las últimas capas para adaptarlas al nuevo problema.
Se realiza un nuevo entrenamiento con los datos específicos de la tarea.

Esto permite ahorrar tiempo y recursos cuando no se dispone de un conjunto de datos muy grande.

6. Detección de objetos

La detección de objetos busca identificar qué objetos aparecen en una imagen y dónde están ubicados.

Dependiendo de la técnica, se pueden obtener:

Cajas delimitadoras (bounding boxes): indican la ubicación de cada objeto mediante un rectángulo.
Máscaras de píxeles: indican exactamente qué píxeles pertenecen a cada objeto.

Una red convolucional puede procesar imágenes y extraer características que posteriormente se utilizan para clasificar y localizar objetos.
