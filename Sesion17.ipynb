**Entrenando redes profundas**
1. El reto de los gradientes
Al entrenar redes neuronales muy profundas aparece el problema de los gradientes inestables. Al propagarse el error hacia las primeras capas, los gradientes pueden:

Desvanecerse: se vuelven cada vez más pequeños y las primeras capas prácticamente dejan de aprender.

Explotar: se vuelven demasiado grandes y provocan actualizaciones inestables.

Un problema relacionado aparece con la función sigmoide, porque en sus extremos se satura: su pendiente es muy pequeña y, por tanto, genera gradientes pequeños.

2. Inicialización de los pesos
Una buena inicialización ayuda a evitar que las señales se hagan demasiado grandes o pequeñas durante la propagación.

La idea principal es mantener una varianza adecuada entre las entradas y las salidas de cada capa.

Un método importante es la inicialización de Glorot (Xavier), especialmente utilizada con funciones como sigmoide y tanh. Busca mantener estable la escala de las señales a través de la red.

3. Funciones de activación
Las funciones de activación permiten que la red aprenda relaciones no lineales.

ReLU: es sencilla y eficiente; devuelve 0 para valores negativos y el propio valor para positivos.

ReLU muerta: ocurre cuando una neurona ReLU queda produciendo constantemente 0 y deja de aprender.

Leaky ReLU: introduce una pequeña pendiente para los valores negativos, reduciendo el problema de las neuronas muertas.

ELU: es una función suave que puede mejorar el aprendizaje.

SELU: está diseñada para favorecer la auto-normalización de las activaciones.

GELU, Swish y Mish: son funciones de activación suaves utilizadas en arquitecturas modernas.

4. Normalización por lotes
La normalización por lotes (Batch Normalization) ayuda a estabilizar y acelerar el entrenamiento.

De forma simplificada, el proceso consiste en:

Calcular estadísticas del lote, como la media y la varianza.

Centrar los valores.

Escalarlos para controlar su dispersión.

Aplicar parámetros aprendibles para desplazar y ajustar las activaciones.

Esto ayuda a que las diferentes capas trabajen con valores más estables.

5. Aprendizaje por transferencia
El aprendizaje transferido (Transfer Learning) consiste en reutilizar una red profunda que ya fue entrenada para una tarea y adaptarla a otra tarea relacionada.

Sus principales ventajas son:

Reduce el tiempo de entrenamiento.

Necesita menos datos.

Ahorra recursos computacionales.

Permite aprovechar características que la red ya aprendió.

Una estrategia común es congelar las capas inferiores o base de la red y entrenar solamente las capas superiores para la nueva tarea.

También se pueden utilizar capas preentrenadas y realizar un ajuste posterior (fine-tuning) de algunas de ellas.

6. Preentrenamiento
El preentrenamiento consiste en entrenar inicialmente una red o modelo con una gran cantidad de datos para que aprenda características generales. Después, ese conocimiento se utiliza para una tarea específica.

En conjunto, las técnicas de buena inicialización, funciones de activación adecuadas, normalización y aprendizaje por transferencia permiten entrenar redes profundas de manera más estable y eficiente.
