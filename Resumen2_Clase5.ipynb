RESUMEN

CRISP-DM (Cross-Industry Standard Process for Data Mining) es una metodología que funciona como un “plano maestro” para realizar proyectos de minería de datos. Su objetivo es convertir un problema de negocio en una solución basada en datos y modelos.

Su desarrollo comenzó alrededor de 1996, posteriormente fue publicado oficialmente en 1999, y con el tiempo surgieron extensiones para adaptarlo a nuevos contextos y tecnologías.

CRISP-DM se caracteriza por ser un ciclo dinámico e iterativo de seis fases, es decir, no necesariamente se avanza una sola vez de principio a fin, sino que se puede regresar a fases anteriores cuando los resultados lo requieren.

Las 6 fases de CRISP-DM
Comprensión del negocio: se identifica el problema, la situación actual, los objetivos y las metas que se quieren alcanzar mediante los datos.
Comprensión de los datos: se recolectan, describen y exploran los datos, además de comprobar su calidad. Aquí es importante preguntarse: ¿qué tan limpios están realmente los datos?
Preparación de los datos: se limpian y transforman los datos para poder utilizarlos. También se realiza la ingeniería de características. Un principio importante es: “basura entra, basura sale”.
Modelado: se seleccionan técnicas y algoritmos, se construyen modelos y se realizan pruebas para encontrar la mejor solución.
Evaluación: se analizan los resultados del modelo y se verifica si realmente cumplen los objetivos del negocio. También se determinan los próximos pasos.
Despliegue: el modelo se lleva a la práctica. Se realiza su implementación, monitoreo, elaboración de informes y revisión de su funcionamiento.
Un ejemplo puede ser la detección de fraude, donde un modelo identifica posibles operaciones fraudulentas. Para evaluar si el modelo es conveniente, se pueden comparar los verdaderos positivos (VP), falsos positivos (FP) y falsos negativos (FN) con el beneficio o costo económico que representa cada caso.

CRISP-DM y la inteligencia artificial
Con la IA moderna, CRISP-DM sigue siendo útil porque proporciona una estructura para organizar los proyectos de datos. Sin embargo, puede complementarse con nuevas técnicas de machine learning, automatización e IA generativa, haciendo que algunas etapas sean más rápidas y automatizadas. El reto seguirá siendo mantener la calidad de los datos, interpretar correctamente los resultados y asegurar que los modelos realmente generen valor para el negocio.
