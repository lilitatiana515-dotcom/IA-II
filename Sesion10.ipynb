Entrenando modelos
1. Regresión lineal
La regresión lineal se utiliza para predecir valores continuos, buscando una relación entre una variable y una o varias características.

Ecuación normal: es una solución matemática directa que calcula los parámetros óptimos para minimizar el error.

Ejemplo: crear un modelo lineal que prediga la satisfacción de vida de un país a partir de su PIB.

2. Método de descenso por gradiente
El descenso por gradiente es un método de optimización que busca encontrar los parámetros que producen el menor error posible.

Sus pasos principales son:

Derivadas: indican hacia qué dirección aumenta o disminuye el error.

Descenso: se modifican los parámetros en la dirección que reduce el error.

Tolerancia: es un número pequeño que permite determinar cuándo los cambios son suficientemente pequeños para detener el proceso.

Tasa de aprendizaje: determina el tamaño de cada paso que realiza el algoritmo.

3. Tipos de descenso por gradiente
Existen tres métodos principales:

Descenso por lotes (Batch): utiliza todos los datos para realizar cada actualización. Es estable, pero puede ser más lento.

Descenso estocástico: utiliza un solo ejemplo en cada paso. Su trayectoria puede "rebotar", pero ese comportamiento puede ayudarle a escapar de mínimos locales.

Descenso por mini lotes: utiliza pequeños grupos de datos. Es un equilibrio entre los métodos por lotes y estocástico y suele ser muy utilizado en la práctica.

Los tres métodos pueden seguir rutas diferentes en el espacio de parámetros hasta llegar a una solución que minimice el error.

4. Regresión polinomial
La regresión polinomial permite trabajar con datos que presentan relaciones más complejas o curvas.

Es un modelo lineal al que se le agregan características elevadas a diferentes potencias, por ejemplo:

x, x², x³...

Esto permite que el modelo se adapte mejor a patrones que no pueden representarse mediante una línea recta.

5. Curvas de aprendizaje
Las curvas de aprendizaje sirven para analizar cómo aprende un modelo y detectar problemas.

Subajuste (underfitting): las curvas se mantienen en niveles de error altos y están muy juntas. Esto indica que el modelo es demasiado simple y no está aprendiendo suficientemente bien.

Sobreajuste (overfitting): existe una gran diferencia entre las curvas de entrenamiento y validación. El modelo aprende muy bien los datos de entrenamiento, pero tiene dificultades con datos nuevos.

6. Regularización
La regularización consiste en aplicar restricciones matemáticas al modelo para evitar que sea demasiado complejo y reducir el sobreajuste.

Algunos métodos son:

Regresión Ridge: agrega una penalización que limita el tamaño de los parámetros.

Regresión Lasso: también penaliza los parámetros y puede hacer que algunos sean exactamente cero.

Parada temprana (Early Stopping): detiene el entrenamiento cuando el error de validación alcanza su mínimo, evitando que el modelo continúe aprendiendo el ruido de los datos.

7. Clasificación
La clasificación busca predecir clases discretas en lugar de valores continuos.

Por ejemplo:

¿Correo spam o no spam?

¿Cliente compra o no compra?

¿Animal gato, perro u otra clase?

Para adaptar la optimización a este tipo de problemas se utilizan modelos como:

Regresión logística: se utiliza principalmente para clasificación binaria y produce una probabilidad entre 0 y 1.

Regresión softmax: permite realizar clasificación multiclase, asignando probabilidades a diferentes clases.
