**Reducción de Dimensionalidad**

- Concepto y Objetivos Principales
Organización y Comprensión: Facilita la visualización y el entendimiento de los datos al simplificar su estructura.

Eliminación de Ruido: Filtra información irrelevante o redundante para mejorar el rendimiento de los modelos.

- La Maldición de la Dimensionalidad
Definición: Ocurre cuando los datos se encuentran en espacios de alta dimensión, provocando que estén extremadamente dispersos.

Consecuencia: Aumenta significativamente el riesgo de sobreajuste (overfitting) debido a la falta de densidad de datos por unidad de espacio.

- Enfoques: Proyección y Aprendizaje de Variedad (Manifold Learning)
Proyección: Proyecta datos de un espacio de alta dimensión a un subespacio de menor dimensión (ej. el conjunto de datos tridimensional tipo brazo de gitano / Swiss Roll).

Aprendizaje de Variedad (Manifold Learning):

Modela la estructura intrínseca y compleja de los datos en espacios no lineales.

Busca conservar la geometría y relaciones locales intactas al desplegar la estructura.

- Análisis de Componentes Principales (PCA)
Objetivo: Encuentra el hiperplano más cercano a los datos y realiza la proyección preservando la máxima varianza posible (la línea de puntos se "aplasta" sobre el hiperplano).

Pasos del algoritmo PCA:

Centrar: Restar la media a los datos.

Aplicar SVD: Descomposición en Valores Singulares (Singular Value Decomposition).

Proyectar: Proyectar los datos en los componentes principales seleccionados.

Curva de Varianza Explicada vs. Número de Dimensiones: Herramienta gráfica para determinar el número óptimo de dimensiones a conservar según el porcentaje de información que se desea mantener.

- Optimizaciones de PCA y Proyección Aleatoria
PCA Incremental: Procesa los datos en minilotes, ideal para conjuntos de datos grandes que no caben en memoria RAM.

PCA Aleatorio (Randomized PCA): Algoritmo estocástico que aproxima rápidamente los primeros componentes principales.

Proyección Aleatoria (Random Projection):

Muy eficiente para datos con dimensiones masivas.

Utiliza una matriz aleatoria y solo requiere la forma/dimensiones de la matriz para funcionar.

- LLE y Otras Técnicas No Lineales
Locally Linear Embedding (LLE): Mantiene las relaciones lineales locales con los vecinos más cercanos, permitiendo "desenrollar" estructuras complejas como el Swiss Roll.

Otras técnicas de reducción:

Escalamiento Multidimensional (MDS): Intenta preservar las distancias entre las instancias.

Isomap: Utiliza distancias geodésicas (distancias sobre la superficie de la variedad) para mantener la geometría global.

t-SNE: Enfocado en mantener las instancias similares agrupadas, ideal para la visualización de datos en 2D o 3D.
