Vectores de Soporte (SVM)
Los SVM (Support Vector Machines) son modelos versátiles de aprendizaje automático que pueden utilizarse para clasificación lineal y no lineal, regresión y detección.

Clasificación lineal: busca una frontera o hiperplano que separe los datos maximizando el margen. Los puntos más cercanos a la frontera son los vectores de soporte.

Margen duro y margen blando: el margen blando permite algunos errores para obtener un modelo más flexible. Se controla mediante el hiperparámetro C:

C bajo: permite más errores y una frontera más flexible.

C alto: penaliza más los errores y busca separar mejor los datos.

Modelos no lineales: cuando los datos no pueden separarse mediante una línea recta, se pueden utilizar características polinomiales.

Truco del kernel: permite trabajar con transformaciones de características sin tener que calcularlas explícitamente. Existen kernels como el polinomial.

Kernel RBF: utiliza una función gaussiana para medir la similitud o distancia entre puntos y crear una separación no lineal.

Hiperparámetro gamma (γ): determina el rango de influencia de cada punto. Los cambios en γ y C modifican la forma y flexibilidad de la frontera.

Regresión SVM: también puede utilizarse para regresión. El parámetro ε (epsilon) define una zona de tolerancia alrededor de la predicción donde los errores no se penalizan.

Matemática interna: SVM busca minimizar los pesos W. En general, pesos más pequeños producen un margen más amplio, ayudando a encontrar una frontera con mayor capacidad de generalización.

En resumen: SVM busca encontrar una frontera que separe o ajuste los datos de la mejor manera posible, utilizando vectores de soporte, márgenes, C, kernels y gamma para controlar cómo aprende el modelo.
