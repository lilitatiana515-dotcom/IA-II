**Clasificación y diagnósticos**

1. Introducción a MNIST
MNIST es un conjunto de datos muy utilizado para aprender y probar modelos de clasificación.

Contiene 70.000 imágenes de dígitos escritos a mano, del 0 al 9.
Cada imagen representa un número, por ejemplo, el 5.
Se considera el “Hola Mundo” del aprendizaje automático, porque es uno de los ejemplos clásicos para comenzar a trabajar con clasificación de imágenes.
El objetivo es que el modelo aprenda a reconocer qué número representa cada imagen.

2. Clasificación binaria
La clasificación binaria consiste en clasificar los datos en solamente dos categorías.

Por ejemplo, con MNIST podríamos crear un modelo que determine:

5 → verdadero
No es 5 → falso
Los principales pasos son:

Definir el objetivo: determinar qué queremos que identifique el modelo.
Crear los datos o vectores: representar las imágenes de manera que el modelo pueda procesarlas.
Seleccionar el modelo: elegir el algoritmo de clasificación adecuado.
Entrenar el modelo: utilizar los datos para que aprenda a realizar la clasificación.
Evaluar el modelo: comprobar qué tan bien clasifica nuevos datos.

3. La trampa de la exactitud
La exactitud (accuracy) indica qué porcentaje de predicciones fueron correctas.

Sin embargo, no siempre es la mejor medida de rendimiento.

Por ejemplo, si tenemos 100 imágenes y solo 5 son números 5, un modelo que diga siempre “no es 5” tendría un 95 % de exactitud, pero en realidad sería inútil para detectar los 5.



4. Matriz de confusión
La matriz de confusión permite observar cómo se comporta un clasificador.

Cuenta cuántas veces:

Una clase fue identificada correctamente.
Una clase fue confundida con otra.
Se produjeron falsos positivos.
Se produjeron falsos negativos.
Por ejemplo, un modelo podría confundir un 5 con un 3.


5. Precisión y exhaustividad
Dos métricas importantes son:

Precisión (Precision): indica la calidad de las predicciones positivas.

De todos los casos que el modelo dijo que eran positivos, ¿cuántos realmente lo eran?

Exhaustividad (Recall): indica qué porcentaje de los casos positivos reales logró encontrar el modelo.

De todos los casos positivos que existían, ¿cuántos encontró?

Existe normalmente un intercambio entre precisión y exhaustividad: aumentar una puede provocar que disminuya la otra.

6. Curva ROC
La curva ROC es una herramienta para analizar el rendimiento de un clasificador.

Compara:

Tasa de verdaderos positivos (TPR) → casos positivos detectados correctamente.
Tasa de falsos positivos (FPR) → casos negativos identificados incorrectamente como positivos.
Sirve para analizar cómo cambia el comportamiento del modelo cuando modificamos el umbral de decisión.

7. Clasificación multiclase
En la clasificación multiclase, el modelo debe elegir entre más de dos categorías.

En MNIST, por ejemplo, existen 10 clases:

0, 1, 2, 3, 4, 5, 6, 7, 8 y 9.

Aquí los errores pueden ser más variados. Por ejemplo:

Un 5 clasificado como 3.
Un 8 clasificado como 9.
Un 7 clasificado como 1.
Para analizar estos errores se puede utilizar una matriz de confusión multiclase.

Una matriz de confusión normalizada permite observar los errores en forma de porcentajes, facilitando la comparación entre las diferentes clases.

8. ¿Priorizar precisión o exhaustividad?
Depende del objetivo del sistema.

* Si es más importante evitar falsas alarmas, se prioriza la precisión.

* Si es más importante encontrar todos los casos positivos, se prioriza la exhaustividad.

Por ejemplo, en un sistema para detectar una enfermedad, normalmente interesa mucho la exhaustividad, porque sería grave que el sistema no detectara a una persona que realmente está enferma.
